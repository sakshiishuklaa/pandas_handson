```python
#data-aggregation
import pandas as pd
df=pd.read_csv('orders.csv')
```

```python
df['sales'].max()#to find highest order
```

```python
df['sales'].min()#to find lowest order
```

```python
df['region'].nunique()#to find unique values
```

```python
df['category'].unique()
```

```python
df.groupby("region")["sales"].sum() #aggregation using group by
```

```python
df.groupby("region")["sales"].agg(["min","max"]) #aggregate and give min and max sales for each region
```

```python
df.groupby("region")["sales"].agg(["min","max","sum"]) #aggregate and give min, max, sum of sales for each region
```

```python
df.groupby("category").agg({'sales':'sum','profit':'sum'}) #another way to use aggregation
```

```python
df["category"].value_counts() #to get the count
```

```python
pd.pivot_table(
    df,
    index="region",
    columns="category",
    values="sales",
    aggfunc="sum",
    fill_value=0
) #creating a excel like pivot table and storing agg values based on columns
```

```python
emp1={"id":[1,2,3],
     "name":["Ankit","Sumit","Amit"],
     "salary":[100,200,300]}
emp2={"id":[4,5,6],
     "name":["Nachiket","Rahul","Vinay"],
     "salary":[400,500,600]}

df1=pd.DataFrame(emp1)
df2=pd.DataFrame(emp2)
```

```python
from sqlalchemy import create_engine, text
engine = create_engine("mysql+pymysql://root:mysql123@127.0.0.1:3306/sys") # to connect to mysql server
```

```python
query = "SELECT * FROM orders where order_id > 10"
df = pd.read_sql(query, con=engine)
```

```python
df.to_sql(name="orders_copy", con=engine, index=False, if_exists="replace")# to write data to mysql
```

```python
df.to_sql(name="orders_copy", con=engine, index=False, if_exists="append") # to append data
```

```python
query = text("truncate table orders_copy)
with engine.begin() as conn:
    conn.execute(query)
    conn.commit()
```
