```python
import pandas as pd
df=pd.read_csv('orders.csv')
df
```

```python
#Data Filtering 
fc=df['region']=='South' #filter condition to based on south region
df.loc[fc]
```

```python
df.loc[df['region']=='South']
```

```python
df.loc[(df['region']=='South') & (df['profit']>0)] #filter condition to based on south region and profit without negative values
#for multiple condition we have to wrap in ()
```

```python
df[df['region']=='South'] # general way 
```

```python
df[~(df['region']=='South')] # to inverse the condition
```

```python
#adding ,updating and deleting columns
```

```python
df['country']='India' #to add new column
```

```python
df['country']='USA' #to update the column
```

```python
df.drop(columns='country')#to drop the column
```

```python
df['profit_flag']=1 #condition to update profit_flag to zero wherever negative
```

```python
df.loc[df['profit']<=0,'profit_flag']=0
df
```

```python
df['profit_by_value'] = None #to add a new column and label them as loss ,low profit and high profit
```

```python
df.loc[df['profit'] < 0, 'profit_by_value'] = 'loss'
```

```python
df.loc[df['profit'].between(0, 50), 'profit_by_value'] = 'low profit'
```

```python
df.loc[df['profit'] > 50, 'profit_by_value'] = 'high profit'
```

```python
df.drop(columns=['profit_flag','status','profit_by_value','country'])
```

```python
df['category_len']=df['category'].apply(len) #to find lenght of category and add under column category_len
```

```python
df
```

```python
def profit_loss(profit):  #creating a func for complex queries
    if profit>0:
        return "profit"
    else:
        return "loss"
df['profit_loss']=df['profit'].apply(profit_loss)
```
