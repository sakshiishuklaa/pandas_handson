```python
import pandas as pd
df=pd.read_csv('orders.csv')
df #empty value represented as NaN in pandas
```

```python
df.isnull()
```

```python
df.isnull().sum() #quick overview of in which columns value is null similar to df.info()
```

```python
df.dropna() #to drop columns wherever null is there or value is missing
```

```python
df.dropna(subset=["category"]) # to drop only selected columns with null values
```

```python
df.fillna("Unknown") # to populate Unknown where NA is there
```

```python
data={
      "name":['Sakshi','Harsh','Rishabh','Rishabh'],
      "age":[25,30,30,30],
     "salary":[25000,30000,35000,35000]
}
df=pd.DataFrame(data) 
df
```

```python
df.drop_duplicates() #to drop exact duplicates by default drops last value
```

```python
df.drop_duplicates(keep='last') #to keep last value
```

```python
df.drop_duplicates(subset='age',keep='first') #to keep first value based on age subset
```

```python
df.drop_duplicates(subset=['age','salary'],keep='first') #to keep first value based on age and salary subset
```

```python
df.dtypes
```

```python
pd.to_datetime(df["order_date"], format='%d-%m-%Y')# to get a series of data in dd-mm-yyyy format
```

```python
df["order_date"]=pd.to_datetime(df["order_date"], format='%d-%m-%Y')
```

```python
df.dtypes #order_date column converted to dd-mm-yyyy format
```

```python
df["ship_date"]=pd.to_datetime(df["ship_date"], format='%d-%m-%Y')
```

```python
df["order_date"]=df["order_date"].dt.year #to get year from date
```
